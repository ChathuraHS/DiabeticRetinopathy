```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

PROJECT = Path(r"D:\DiabeticRetinopathy")

train_df = pd.read_csv(PROJECT / "data/splits/train.csv")
val_df = pd.read_csv(PROJECT / "data/splits/validation.csv")
test_df = pd.read_csv(PROJECT / "data/splits/test.csv")

print("Training:", len(train_df))
print("Validation:", len(val_df))
print("Testing:", len(test_df))

print("\nTraining distribution:")
print(train_df["level"].value_counts().sort_index())
```

```python
class_names = [
    "No DR",
    "Mild",
    "Moderate",
    "Severe",
    "Proliferative"
]

counts = train_df["level"].value_counts().sort_index()

plt.figure(figsize=(9, 5))

sns.barplot(
    x=class_names,
    y=counts.values
)

plt.title("Training Dataset Class Distribution")
plt.xlabel("Disease Severity")
plt.ylabel("Number of Images")
plt.xticks(rotation=20)

plt.tight_layout()
plt.savefig(PROJECT / "reports/class_distribution.png")
plt.show()
```

```python
from PIL import Image

fig, axes = plt.subplots(5, 5, figsize=(15, 15))

for grade in range(5):

    samples = train_df[
        train_df["level"] == grade
    ].sample(
        min(5, len(train_df[train_df["level"] == grade])),
        random_state=42
    )

    for j, (_, row) in enumerate(samples.iterrows()):

        image = Image.open(row["processed_path"])

        axes[grade, j].imshow(image)
        axes[grade, j].axis("off")

        if j == 0:
            axes[grade, j].set_title(class_names[grade])

plt.tight_layout()
plt.savefig(PROJECT / "reports/sample_images.png")
plt.show()
```

```python
split_data = pd.concat([
    train_df.assign(split="Training"),
    val_df.assign(split="Validation"),
    test_df.assign(split="Testing")
])

plt.figure(figsize=(10, 5))

sns.countplot(
    data=split_data,
    x="level",
    hue="split"
)

plt.title("Class Distribution Across Dataset Splits")
plt.xlabel("Disease Grade")
plt.ylabel("Number of Images")

plt.tight_layout()
plt.savefig(PROJECT / "reports/split_distribution.png")
plt.show()
```
