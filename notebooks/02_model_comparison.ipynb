```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

PROJECT = Path(r"D:\DiabeticRetinopathy")
REPORTS = PROJECT / "reports"

MODELS = {
    "ResNet18": "resnet18",
    "EfficientNet-B0": "efficientnet_b0",
}

METRIC_ORDER = [
    "Quadratic Weighted Kappa",
    "QWK (default thresholds)",
    "Accuracy",
    "Balanced Accuracy",
    "Macro Precision",
    "Macro Recall",
    "Macro F1",
]

comparison = (
    pd.concat(
        [pd.read_csv(REPORTS / f"{file}_metrics.csv").assign(Model=name)
         for name, file in MODELS.items()],
        ignore_index=True,
    )
    .set_index("Model").T
    .reindex(METRIC_ORDER)
)

print(comparison.round(4))
comparison.to_csv(REPORTS / "model_comparison.csv")

# Bar chart (QWK first, since it is the target metric)
ax = comparison.plot(kind="bar", figsize=(11, 5), width=0.75)
ax.set_ylabel("Score")
ax.set_ylim(0, 1)
ax.set_title("Test-set model comparison")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.savefig(REPORTS / "model_comparison.png", dpi=150)
plt.show()
```
